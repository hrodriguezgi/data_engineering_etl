# Guided Practice - Error Handling and Logging

This notebook complements `05_error_handling.py`.

Goals:
- Parse values safely with `try/except`
- Keep processing when records fail
- Log processing summary

```python
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s | %(message)s")
logger = logging.getLogger(__name__)

records = [
    {"id": 1, "amount": "250"},
    {"id": 2, "amount": "-10"},
    {"id": 3, "amount": "abc"},
    {"id": 4, "amount": None},
    {"id": 5, "amount": "30.5"},
]
```

## 1) Safe Parser

```python
def safe_parse_amount(value):
    try:
        amount = float(value)
    except (TypeError, ValueError):
        return None
    if amount <= 0:
        return None
    return amount
```

## 2) Process Records with Error Collection

```python
valid, errors = [], []
for row in records:
    amount = safe_parse_amount(row.get("amount"))
    if amount is None:
        errors.append({"id": row["id"], "reason": f"Invalid amount: {row.get('amount')!r}"})
        continue
    valid.append({"id": row["id"], "amount": amount})

summary = {
    "valid_count": len(valid),
    "invalid_count": len(errors),
    "errors": errors,
}

logger.info("valid=%s invalid=%s", summary["valid_count"], summary["invalid_count"])
summary
```
