Goals:
- Practice `if/elif/else`
- Use `for`/`while` loops
- Route records with conditions

```python
records = [
    {"id": 1, "status": "active", "amount": 250},
    {"id": 2, "status": "inactive", "amount": 90},
    {"id": 3, "status": "active", "amount": 175},
    {"id": 4, "status": "active", "amount": 320},
]
records
```

## 1) Conditionals

```python
def classify_amount(amount):
    if amount >= 300:
        return "high"
    elif amount >= 150:
        return "medium"
    return "low"


[classify_amount(r["amount"]) for r in records]
```

## 2) For Loop Routing

```python
priority, standard = [], []
for row in records:
    if row["status"] != "active":
        continue
    if row["amount"] >= 200:
        priority.append(row)
    else:
        standard.append(row)

len(priority), len(standard)
```

## 3) While Loop (Batch Processing)

```python
batch_size = 2
i = 0
batches = []
while i < len(records):
    batches.append(records[i : i + batch_size])
    i += batch_size

[len(b) for b in batches]
```
