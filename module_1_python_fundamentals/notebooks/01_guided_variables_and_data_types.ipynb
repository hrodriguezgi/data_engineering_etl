## 4) Type Conversion

## 5) Mini Data Record Summary

```python
records = [
    {"id": 1, "customer": "Alice", "amount": "250.00"},
    {"id": 2, "customer": "Bob", "amount": "-20"},
    {"id": 3, "customer": "", "amount": "175.5"},
    {"id": 4, "customer": "Carol", "amount": "120.0"},
]

valid = []
for row in records:
    customer = row["customer"].strip()
    try:
        amount = float(row["amount"])
    except ValueError:
        continue

    if not customer or amount <= 0:
        continue

    valid.append({"id": row["id"], "customer": customer, "amount": amount})

summary = {
    "valid_count": len(valid),
    "total_amount": round(sum(r["amount"] for r in valid), 2),
    "customers": sorted([r["customer"] for r in valid]),
}

summary
```
