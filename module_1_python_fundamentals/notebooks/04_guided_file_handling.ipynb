# Guided Practice - File Handling

This notebook complements `04_file_handling.py`.

Goals:
- Read CSV data
- Transform rows
- Write CSV and JSON summary

```python
from pathlib import Path
import csv
import json

base_dir = Path(".")
input_path = base_dir / "sales_notebook_input.csv"

sample = [
    ["order_id", "customer", "quantity", "unit_price"],
    ["1001", "Alice", "1", "999.99"],
    ["1002", "Bob", "3", "29.99"],
    ["1003", "Carol", "2", "140.00"],
]

with open(input_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(sample)

input_path
```

## 1) Read and Transform CSV

```python
with open(input_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

transformed = []
for row in rows:
    quantity = int(row["quantity"])
    unit_price = float(row["unit_price"])
    total = round(quantity * unit_price, 2)
    transformed.append(
        {
            "order_id": row["order_id"],
            "customer": row["customer"],
            "quantity": quantity,
            "unit_price": unit_price,
            "total": total,
        }
    )

transformed
```

## 2) Write Output CSV + JSON Summary

```python
output_csv = base_dir / "sales_notebook_output.csv"
summary_json = base_dir / "sales_notebook_summary.json"

with open(output_csv, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["order_id", "customer", "quantity", "unit_price", "total"])
    writer.writeheader()
    writer.writerows(transformed)

summary = {
    "rows_processed": len(transformed),
    "grand_total": round(sum(r["total"] for r in transformed), 2),
}

with open(summary_json, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

summary
```
