Goals:
- Build reusable ETL functions
- Pass arguments and defaults
- Compose functions into a pipeline

```python
raw_records = [
    {"id": 1, "status": "active", "amount": "250.00"},
    {"id": 2, "status": "inactive", "amount": "90"},
    {"id": 3, "status": "active", "amount": "175.5"},
    {"id": 4, "status": "active", "amount": "invalid"},
]
```

## 1) Define ETL Functions

```python
def extract_active(records):
    return [r for r in records if r["status"] == "active"]


def transform_amount(records, tax_rate=0.08):
    out = []
    for row in records:
        try:
            amount = float(row["amount"])
        except (TypeError, ValueError):
            continue
        new_row = row.copy()
        new_row["amount"] = amount
        new_row["amount_with_tax"] = round(amount * (1 + tax_rate), 2)
        out.append(new_row)
    return out


def summarize(records):
    return {
        "count": len(records),
        "total_amount": round(sum(r["amount"] for r in records), 2),
        "total_amount_with_tax": round(sum(r["amount_with_tax"] for r in records), 2),
    }
```

## 2) Run Pipeline

```python
active = extract_active(raw_records)
transformed = transform_amount(active, tax_rate=0.08)
summary = summarize(transformed)
summary
```
